# Compare primera vuelta files: 2026 (xlsx) vs 2022 (csv)

Opens the two `primera_vuelta` result files and compares their dimensions:

- `primera_vuelta_2026/MUNICIPIO_PRESIDENTE_31-05-2026.xlsx` - the only Excel file inside a `primera_vuelta` folder.
- `primera_vuelta_mesas_2022/muns_votes_2022_1v.csv` - 2022 first-round data (CSV, not Excel).

```python
from pathlib import Path

import pandas as pd

RAW_DIR = Path.cwd()

xlsx_path = RAW_DIR / "primera_vuelta_2026" / "MUNICIPIO_PRESIDENTE_31-05-2026.xlsx"
csv_path = RAW_DIR / "primera_vuelta_mesas_2022" / "muns_votes_2022_1v.csv"

votes_2026 = pd.read_excel(xlsx_path)
votes_2022 = pd.read_csv(csv_path)

print(f"2026 (xlsx): {xlsx_path.name} -> shape {votes_2026.shape}")
print(f"2022 (csv):  {csv_path.name} -> shape {votes_2022.shape}")
```

```python
# Compare dimensions directly: rows, columns, and the shared identifier columns.
shared_columns = sorted(set(votes_2026.columns) & set(votes_2022.columns))

comparison = pd.DataFrame(
    {
        "file": ["MUNICIPIO_PRESIDENTE_31-05-2026.xlsx (2026)", "muns_votes_2022_1v.csv (2022)"],
        "n_rows": [votes_2026.shape[0], votes_2022.shape[0]],
        "n_columns": [votes_2026.shape[1], votes_2022.shape[1]],
    }
)

print(comparison.to_string(index=False))
print()
print(f"Row count difference (2026 - 2022): {votes_2026.shape[0] - votes_2022.shape[0]}")
print(f"Column count difference (2026 - 2022): {votes_2026.shape[1] - votes_2022.shape[1]}")
print()

```

```python
print("2026 columns:", list(votes_2026.columns))
print(votes_2026.head(10).to_string())
```

```python
votes_2022["NUM_AÑO"] = 2022
votes_2022 = votes_2022.rename(columns={"votes": "NUM_VOT"})
print("2022 columns:", list(votes_2022.columns))
print(votes_2022.head(2).to_string())
```

```python
print(f"Shared column names ({len(shared_columns)}): {shared_columns}")
print(f"2026-only columns ({votes_2026.shape[1] - len(shared_columns)}): "
      f"{sorted(set(votes_2026.columns) - set(votes_2022.columns))}")
print(f"2022-only columns ({votes_2022.shape[1] - len(shared_columns)}): "
      f"{sorted(set(votes_2022.columns) - set(votes_2026.columns))}")
```

```python
SHARED_COLUMNS = ["COD_DDE", "COD_MME", "DES_CAN", "DES_DD", "DES_MM", "NUM_AÑO"]

```

```python
# Municipio x candidate level: one row per (municipio, candidate) combo, matching on the
# numeric codes (COD_DDE, COD_MME - not the DES_DD/DES_MM text, which has a few spelling
# inconsistencies between files) plus DES_CAN. Outer join, since most 2022 and 2026
# candidates are different people/ballot lines - only "VOTOS EN BLANCO", "VOTOS NO
# MARCADOS", and "VOTOS NULOS" are shared DES_CAN values across the two years, so most
# rows will only have a votes_2026 or a votes_2022 value, not both.
ROW_COLUMNS = ["COD_DDE", "DES_DD", "COD_MME", "DES_MM", "DES_CAN"]
MERGE_KEYS = ["COD_DDE", "COD_MME", "DES_CAN"]

candidates_2026 = votes_2026[ROW_COLUMNS + ["NUM_VOT"]]
candidates_2022 = votes_2022[ROW_COLUMNS + ["NUM_VOT"]]

candidate_merged = candidates_2026.merge(
    candidates_2022.drop(columns=["DES_DD", "DES_MM"]),
    on=MERGE_KEYS, how="outer", suffixes=("_2026", "_2022"),
)

both_years = candidate_merged["NUM_VOT_2026"].notna() & candidate_merged["NUM_VOT_2022"].notna()
print(f"2026 rows: {len(candidates_2026)}, 2022 rows: {len(candidates_2022)}, "
      f"merged (1 row per municipio+candidate): {len(candidate_merged)}")
print(f"Rows with a value in both years: {both_years.sum()} "
      f"(only VOTOS EN BLANCO/NO MARCADOS/NULOS, since real candidates differ by year)")
print(candidate_merged.head(3).to_string())
print()
print(candidate_merged[both_years].head(3).to_string())
```

```python
# Clean up display: DES_CAN from ALL CAPS to Title Case, and the vote columns to a proper
# integer dtype. Nullable "Int64" is used (not plain int) since the outer join above left
# NaN where a candidate only ran in one of the two years.
candidate_merged["DES_CAN"] = candidate_merged["DES_CAN"].str.title()
candidate_merged["NUM_VOT_2026"] = candidate_merged["NUM_VOT_2026"].astype("Int64")
candidate_merged["NUM_VOT_2022"] = candidate_merged["NUM_VOT_2022"].astype("Int64")

print(candidate_merged.dtypes)
print(candidate_merged.head(3).to_string())
```

```python
candidates_2026 = sorted(candidate_merged["DES_CAN"].dropna().unique())

print(f"2022 and 2026 unique DES_CAN ({len(candidates_2026)}):")
for name in candidates_2026:
    print(f"  {name}")
```

```python
# Vote share by candidate, one panel per year (2022 on top, 2026 below), sorted by share
# descending, plus a blank column for manual notes. Printed only.
def candidate_share_panel(year, votes_df, candidate_col, vote_col):
    totals = votes_df.groupby(candidate_col)[vote_col].sum()
    shares = (totals / totals.sum() * 100).sort_values(ascending=False)

    lines = [
        f"  \\textbf{{{year}}}\\\\[2pt]",
        "  \\begin{tabular}{l | r | l}",
        "  \\toprule",
        "  \\textbf{Candidate} & \\textbf{Votes (\\%)} & \\phantom{XXXXXXXXXX} \\\\",
        "  \\midrule",
    ]
    for name, pct in shares.items():
        lines.append(f"  {name.title()} & {pct:.1f}\\,\\% & \\\\")
    lines.append("  \\bottomrule")
    lines.append("  \\end{tabular}")
    return "\n".join(lines)


table_lines = [
    r"\begin{table}[htbp]",
    r"  \centering",
    r"  \tiny",
    r"  \caption{Vote share by candidate, 2022 and 2026 first-round presidential elections}",
    candidate_share_panel(2022, votes_2022, "DES_CAN", "NUM_VOT"),
    r"  \vspace{6pt}",
    candidate_share_panel(2026, votes_2026, "DES_CAN", "NUM_VOT"),
    r"  \label{tab:candidate_vote_share_2022_2026}",
    r"\end{table}",
]

print("\n".join(table_lines))
```
